```python
arrivals_df = spark.table("workspace.default.bronze_arrivals")
```

```python
from pyspark.sql.functions import *

silver_arrivals = (
    arrivals_df
    .withColumn("arrival_record_id", trim(col("arrival_record_id")))
    .withColumn("market_id", upper(trim(col("market_id"))))
    .withColumn("commodity_id", upper(trim(col("commodity_id"))))
    .withColumn("variety_id", upper(trim(col("variety_id"))))
   .withColumn(
    "report_date",
    expr("""
        COALESCE(
            CAST(try_to_timestamp(report_date,'yyyy-MM-dd') AS DATE),
            CAST(try_to_timestamp(report_date,'dd-MM-yyyy') AS DATE)
        )
    """)
)
    .withColumn("arrival_unit", upper(trim(col("arrival_unit"))))
    .withColumn("source_system", upper(trim(col("source_system"))))
    .withColumn("source_file_version", trim(col("source_file_version")))
    .withColumn("load_batch_id", trim(col("load_batch_id")))
    .withColumn("raw_market_name", upper(trim(col("raw_market_name"))))
    .withColumn("raw_commodity_name", upper(trim(col("raw_commodity_name"))))
    .withColumn("raw_variety_name", upper(trim(col("raw_variety_name"))))
    .withColumn("source_file", trim(col("source_file")))
)
```

```python
arrivals_df.select("report_date").show(20, False)
```

```python
silver_arrivals.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable("workspace.default.silver_arrivals_candidate")
```

```python
from pyspark.sql.functions import *

prices_df = spark.table("workspace.default.bronze_prices")
```

```python
silver_prices = (
    prices_df

    .withColumn("price_record_id", trim(col("price_record_id")))
    .withColumn("market_id", upper(trim(col("market_id"))))
    .withColumn("commodity_id", upper(trim(col("commodity_id"))))
    .withColumn("variety_id", upper(trim(col("variety_id"))))

    .withColumn(
        "report_date",
        expr("CAST(try_to_timestamp(report_date,'yyyy-MM-dd') AS DATE)")
    )

    .withColumn("price_unit", upper(trim(col("price_unit"))))
    .withColumn("source_system", upper(trim(col("source_system"))))
    .withColumn("source_file_version", trim(col("source_file_version")))
    .withColumn("load_batch_id", trim(col("load_batch_id")))
    .withColumn("raw_market_name", upper(trim(col("raw_market_name"))))
    .withColumn("raw_commodity_name", upper(trim(col("raw_commodity_name"))))
    .withColumn("raw_variety_name", upper(trim(col("raw_variety_name"))))
    .withColumn("source_file", trim(col("source_file")))

    .filter(col("report_date").isNotNull())
)
```

```python
silver_prices.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable("workspace.default.silver_prices_candidate")
```

```python
from pyspark.sql import functions as F

print("=" * 75)
print("🌾 AGRIPULSE — SILVER TRANSFORMATION EVIDENCE")
print("=" * 75)

# Get all tables from workspace.default
tables_df = spark.sql("SHOW TABLES IN workspace.default")

# Find tables containing "silver" in their name
silver_tables = (
    tables_df
    .filter(F.lower(F.col("tableName")).contains("silver"))
    .select("tableName")
    .collect()
)

if len(silver_tables) == 0:

    print("\n❌ No Silver tables found in workspace.default")
    print("\nAvailable tables are:")

    tables_df.select("tableName").show(truncate=False)

else:

    print(f"\nFound {len(silver_tables)} Silver table(s):\n")

    for row in silver_tables:

        table_name = row["tableName"]
        full_table_name = f"workspace.default.{table_name}"

        df = spark.table(full_table_name)

        print("=" * 75)
        print(f"📋 TABLE: {table_name}")
        print("=" * 75)

        print(f"Rows    : {df.count():,}")
        print(f"Columns : {len(df.columns)}")
        print(f"Fields  : {', '.join(df.columns)}")

print("\n" + "=" * 75)
print("✅ Silver transformation evidence completed.")
print("=" * 75)
```
