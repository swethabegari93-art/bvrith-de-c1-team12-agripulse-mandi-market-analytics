# AgriPulse Week3B Data Exploration

```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.getOrCreate()
```

```python
display(dbutils.fs.ls("/Volumes/agripulse/default/agri_raw_files"))
```

```python
arrivals=spark.read.option("header",True).csv("/Volumes/agripulse/default/agri_raw_files/arrivals.csv")
display(arrivals)
arrivals.printSchema()
arrivals.createOrReplaceTempView("arrivals")
```

```python
commodities=spark.read.option("header",True).csv("/Volumes/agripulse/default/agri_raw_files/commodities.csv")
display(commodities)
commodities.printSchema()
commodities.createOrReplaceTempView("commodities")
```

```python
markets=spark.read.option("header",True).csv("/Volumes/agripulse/default/agri_raw_files/markets.csv")
display(markets)
markets.printSchema()
markets.createOrReplaceTempView("markets")
```

```python
prices=spark.read.option("header",True).csv("/Volumes/agripulse/default/agri_raw_files/prices.csv")
display(prices)
prices.printSchema()
prices.createOrReplaceTempView("prices")
```

```python
spark.sql("SELECT COUNT(*) FROM arrivals").show()
```
