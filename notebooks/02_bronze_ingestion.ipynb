```python
arrivals_df = (
    spark.read
         .option("header", True)
         .option("inferSchema", True)
         .csv(volume_path + "arrivals.csv")
)
```

```python
display(arrivals_df)
```

```python
arrivals_df.printSchema()
```

```python
print(arrivals_df.count())
```

```python
from pyspark.sql.functions import current_timestamp, lit

bronze_arrivals = (
    arrivals_df
        .withColumn("source_file", lit("arrivals.csv"))
        .withColumn("ingestion_time", current_timestamp())
)
```

```python
bronze_arrivals.write \
    .format("delta") \
    .mode("overwrite") \
    .save("/Volumes/agripulse/default/agri_raw_files/bronze_arrivals")
```

```python
display(
    spark.read.format("delta")
         .load("/Volumes/agripulse/default/agri_raw_files/bronze_arrivals")
)
```
